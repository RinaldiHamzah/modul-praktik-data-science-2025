Cleaning

```python
#mengahpus angka nul
df['pizza_id'] = df['pizza_id'].astype(int)
df['order_id'] = df['order_id'].astype(int)
df['quantity'] = df['quantity'].astype(int)
```

```python
df['unit_price'] = df['unit_price'].astype(int)
df['total_price'] = df['total_price'].astype(int)
```

```python
df['order_date'] = pd.to_datetime(
    df['order_date'],
    format='mixed',
    dayfirst=True
)
```

```python
df['order_time'] = pd.to_datetime(
    df['order_time'],
    format='mixed'
).dt.time
```

```python
df['pizza_size'] = df['pizza_size'].str.lower().str.strip()
df['pizza_category'] = df['pizza_category'].str.lower().str.strip()
df['pizza_name'] = df['pizza_name'].str.lower().str.strip()
df['pizza_ingredients'] = (
    df['pizza_ingredients']
    .str.lower()
    .str.replace('[^a-z, ]', '', regex=True)
    .str.strip()
)
```

```python
df['price_check'] = df['quantity'] * df['unit_price']

(df['price_check'] == df['total_price']).value_counts()
```

# 2.2 Filtering, grouping, dan aggregation

FILTERING

```python
#filter berdasarkan ukuran pizza l
df_l = df[df['pizza_size'] == 'l']
df_l.head()
```

```python
#filter penjualan berdasarkan tahun
df_2015 = df[df['order_date'].dt.year == 2015]
df_2015.head()
```

```python
#Filter transaksi dengan quantity
df_qty = df[df['quantity'] > 3]
df_qty.head()
```

GROUPING (Pengelompokan Data)

```python
#Total revenue per kategori pizza
revenue_category = df.groupby('pizza_category')['total_price'].sum()
revenue_category
```

```python
#Total jumlah pizza terjual per ukuran
quantity_size = df.groupby('pizza_size')['quantity'].sum()
quantity_size
```

```python
#Jumlah order per tanggal
orders_per_day = df.groupby('order_date')['order_id'].nunique()
orders_per_day
```

```python
#Penjualan per jam
df['hour'] = pd.to_datetime(df['order_time'].astype(str)).dt.hour

sales_by_hour = df.groupby('hour')['quantity'].sum()
sales_by_hour
```

```python
#Jam dengan penjualan tertinggi
peak_hour = sales_by_hour.idxmax()
peak_value = sales_by_hour.max()
peak_hour, peak_value
```

AGGREGATION (Agregasi Statistik)

```python
#Agregasi lengkap per kategori
summary_category = df.groupby('pizza_category').agg(
    total_revenue=('total_price', 'sum'),
    total_quantity=('quantity', 'sum'),
    avg_unit_price=('unit_price', 'mean'),
    total_orders=('order_id', 'nunique')
)
summary_category
```

```python
#Top 5 pizza paling laku
top_5_pizza = (
    df.groupby('pizza_name')['quantity']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)
top_5_pizza
```

```python
#Best Seller Berdasarkan Revenue
best_revenue = (
    df.groupby('pizza_name')['total_price']
    .sum()
    .sort_values(ascending=False)
    .head(5)
)
best_revenue
```

```python
#Average Order Value (AOV)
aov = df.groupby('order_id')['total_price'].sum().mean()
aov
```

# 2.4 Penanganan data hilang (missing values)

```python
df.isna().sum()
```

```python
#menghapus tabel
df = df.dropna(subset=['price_check'])
```

```python
#Kolom Numerik Transaksi
df['quantity'].fillna(1, inplace=True)
```

```python
df['unit_price'] = df.groupby('pizza_name_id')['unit_price']\
                      .transform(lambda x: x.fillna(x.mean()))
```

```python
df['total_price'] = df['quantity'] * df['unit_price']
```

```python
df['pizza_size'].fillna(df['pizza_size'].mode()[0], inplace=True)
df['pizza_category'].fillna(df['pizza_category'].mode()[0], inplace=True)
```

```python
#Kolom Teks
df['pizza_ingredients'].fillna('Unknown', inplace=True)
df['pizza_name'].fillna('Unknown', inplace=True)
```

1.	Analisis dataset Pizza Sales untuk menentukan jenis dengan jumlah penjualan tertinggi dan tren harian penjualan.

```python
#Jenis Pizza dengan Penjualan Tertinggi
top_quantity = (
    df.groupby('pizza_name')['quantity']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)
print(top_quantity)
```

```python
#Berdasarkan Total Pendapatan
top_revenue = (
    df.groupby('pizza_name')['total_price']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)
print(top_revenue)
```

```python
#Tren Penjualan Harian
daily_sales = (
    df.groupby('order_date')['total_price']
      .sum()
      .reset_index()
)
```

```python
daily_sales['day_name'] = daily_sales['order_date'].dt.day_name()
```

```python
#Rata-rata Penjualan per Hari
day_trend = (
    daily_sales.groupby('day_name')['total_price']
      .mean()
      .reindex([
          'Monday','Tuesday','Wednesday',
          'Thursday','Friday','Saturday','Sunday'
      ])
)

print(day_trend)
```

2.	Lakukan normalisasi terhadap kolom numerik menggunakan NumPy.

```python
#Implementasi dengan NumPy
# Pilih kolom numerik
numerical_cols = ['quantity', 'unit_price', 'total_price']

# Salin data agar aman
df_norm = df.copy()

for col in numerical_cols:
    col_min = np.min(df[col])
    col_max = np.max(df[col])

    df_norm[col] = (df[col] - col_min) / (col_max - col_min)
```

```python
df_norm[numerical_cols].head()
```

```python
for col in numerical_cols:
    mean = np.mean(df[col])
    std = np.std(df[col])

    df_norm[col] = (df[col] - mean) / std
```
